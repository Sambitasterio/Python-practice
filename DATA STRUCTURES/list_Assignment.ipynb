### Assignment 6: List Methods

Create a list of random numbers and sort it in ascending and descending order. Remove the duplicates from the list and print the modified list.

```python
import random

random_numbers = [random.randint(1,20) for _ in range(15)]
print(f"Original List : {random_numbers}")

sorted_numbers = sorted(random_numbers)
print(f"Sorted Numbers : {sorted_numbers}")

reverse_numbers = sorted(random_numbers, reverse = True)
print(f"Reverse Sorted Numbers: {reverse_numbers}")

unique_numbers = list(set(random_numbers))
print(f"The unique numbers: {unique_numbers}")

```

### Assignment 9: Matrix Transposition

Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.

```python
def transpose_matrix(matrix):
    transposed = [[matrix[j][i] for j in range(len(matrix))] for i in range(len(matrix[0]))]
    return transposed

matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

transposed = transpose_matrix(matrix)

print(f"The original matrix :")
for row in matrix:
    print(row)
print(f"The new transposed matrix:")
for row in transposed:
    print(row)
```

### Assignment 10: Flattening a Nested List

Write a function that takes a nested list and flattens it into a single list. Print the original and flattened lists.

```python
def flatend_list(nested_list):
    flat = [item for sublist in nested_list for item in sublist]
    return flat

nested_list = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
flat = flatend_list(nested_list)
print(f"Original nested list")
print(nested_list)
print(f"Flatened list:")
print(flat)
```
