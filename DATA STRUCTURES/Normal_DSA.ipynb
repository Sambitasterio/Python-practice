## 1.Celsius To Fahrenheit

```python
def celcius_to_fahrenheit(C):
    Fahrenheit = (9/5 * C) + 32
    return Fahrenheit

C = float(input("Enter the Temperature in Celcius : "))
result  = celcius_to_fahrenheit(C)
print(f"The {C} c is {result} in Fahrenheit")
```

## 2.Area of a Rectangle

```python
def area_of_rectangle(length,breadth):
    area = length * breadth
    return area
length = int(input("Enter the Length:"))
breadth = int(input("Enter the breadth"))
result = area_of_rectangle(length,breadth)
print(f"The area is {result}")
```

## 3.Number of Rounds of Lift

```python
def calculate_lift_rounds(n, capacity):
    rounds = n//capacity

    if n % capacity !=0:
        rounds+=1
    return rounds
n = int(input("Enter number of people:"))
capacity = int(input("Maximum capacity single ride:"))
result = calculate_lift_rounds(n,capacity)
print(f"The no of rounds: {result}")
```

## 4.Line Equation

```python
def calculate_y(slope,intercept,x):
    y = slope*x + intercept
    return y
slope = int(input("Enter the slope"))
intercept = int(input("Enter the intercept"))
x = int(input("Enter the value"))
result = calculate_y(slope,intercept,x)
print(f"{result}")

```

```python
def check_unique(lst):
    return len(lst) == len(set(lst))
```
