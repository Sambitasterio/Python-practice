## 1.Celsius To Fahrenheit

```python
def celcius_to_fahrenheit(C):
    Fahrenheit = (9/5 * C) + 32
    return Fahrenheit

C = float(input("Enter the Temperature in Celcius : "))
result  = celcius_to_fahrenheit(C)
print(f"The {C} c is {result} in Fahrenheit")
```

## 2.Area of a Rectangle

```python
def area_of_rectangle(length,breadth):
    area = length * breadth
    return area
length = int(input("Enter the Length:"))
breadth = int(input("Enter the breadth"))
result = area_of_rectangle(length,breadth)
print(f"The area is {result}")
```

## 3.Number of Rounds of Lift

```python
def calculate_lift_rounds(n, capacity):
    rounds = n//capacity

    if n % capacity !=0:
        rounds+=1
    return rounds
n = int(input("Enter number of people:"))
capacity = int(input("Maximum capacity single ride:"))
result = calculate_lift_rounds(n,capacity)
print(f"The no of rounds: {result}")
```

## 8.Maximum Difference between 2 consecutive elements in a List

```python
lst1 = [1, 7, 3, 10, 5]
def max_consecutive_diff(lst1):
    if(len(lst1)<2):
        return 0
    max_diff = 0
    for i in range(len(lst1) - 1):
        diff = abs(lst1[i] - lst1[i+1])
        if max_diff < diff:
            max_diff = diff
    return max_diff
result = max_consecutive_diff(lst1)
print(result)
```

```python
##optimal using zip

def max_consecutive_diff(lst1):
    return max(abs(a - b) for a, b in zip(lst1, lst1[1:]))
result = max_consecutive_diff(lst1)
print(result)
```

## 9.Merge two sorted List

```python
def merge_two_sorted_lists(list1, list2):
    # Your code goes here
    result = []
    i,j = 0,0
    while i < len(list1) and j < len(list2):
        if list1[i] <= list2[j]:
            result.append(list1[i])
            i+=1
        else:
            result.append(list2[j])
            j+=1
    
    result.extend(list1[i:])
    result.extend(list2[j:])
    return result
```

## 10.Two Sum
Given an array of integers and a target, return indices of two numbers that add up to the target.

```python
def two_sum(nums,target):
    seen = {}
    for i,n in enumerate(nums):
        diff = target - n
        if diff in seen:
            return[seen[diff],i]
        seen[n] = i
```

## 11.Maximum Subarray (Kadane's)
Find the contiguous subarray with the largest sum.


```python
def max_subarray(nums):
    max_sum = curr = nums[0]
    for n in nums[1:]:
        curr = max(n,curr+n)
        max_sum = max(max_sum,curr)
    return max_sum
```
