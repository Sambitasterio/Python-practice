```python
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
print(fruits[1:])
print(fruits[1:3])#so it will print from 1 to 2 index but not 3
print(fruits[:3])# it will print from 0 to 2 index but not 3
```

# Modifying the List

```python
fruits[1] = "mango"
print(fruits)
```

```python
fruits.append("GRAPES")
del fruits[5]
fruits.remove("GRAPES")
print(fruits)
```

```python
## remove and return the last item of the list
popped_fruit = fruits.pop()
print(popped_fruit)
```
